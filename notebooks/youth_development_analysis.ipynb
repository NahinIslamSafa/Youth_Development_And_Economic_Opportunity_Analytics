```python
# ==============================
# STEP 1 — Prepare GDP
# ==============================

gdp_clean = gdp[
    [
        "REF_AREA",
        "REF_AREA_LABEL",
        "TIME_PERIOD",
        "OBS_VALUE"
    ]
]

gdp_clean = gdp_clean.rename(columns={
    "REF_AREA": "Country_Code",
    "REF_AREA_LABEL": "Country",
    "TIME_PERIOD": "Year",
    "OBS_VALUE": "GDP_per_capita_PPP"
})

print("GDP clean shape:", gdp_clean.shape)
print(gdp_clean.head())


# ==============================
# STEP 2 — Prepare Education
# ==============================

education_clean = education[
    [
        "REF_AREA",
        "REF_AREA_LABEL",
        "TIME_PERIOD",
        "OBS_VALUE"
    ]
]

education_clean = education_clean.rename(columns={
    "REF_AREA": "Country_Code",
    "REF_AREA_LABEL": "Country",
    "TIME_PERIOD": "Year",
    "OBS_VALUE": "Lower_secondary_completion"
})

print("\nEducation clean shape:", education_clean.shape)
print(education_clean.head())
```

```python
# ==============================
# STEP 3 — Prepare NEET
# ==============================

neet_clean = neet[
    neet["sex.label"] == "Total"
][
    [
        "ref_area.label",
        "time",
        "obs_value",
        "source.label",
        "obs_status.label"
    ]
]

neet_clean = neet_clean.rename(columns={
    "ref_area.label": "Country",
    "time": "Year",
    "obs_value": "NEET_Rate",
    "source.label": "NEET_Source",
    "obs_status.label": "NEET_Status"
})

print("\nNEET clean shape:", neet_clean.shape)
print(neet_clean.head())
```

```python
# Check that each dataset still has exactly
# one observation per country-year

print(
    "GDP duplicate country-year rows:",
    gdp_clean.duplicated(["Country_Code", "Year"]).sum()
)

print(
    "Education duplicate country-year rows:",
    education_clean.duplicated(["Country_Code", "Year"]).sum()
)

print(
    "NEET duplicate country-year rows:",
    neet_clean.duplicated(["Country", "Year"]).sum()
)
```

```python
print("Youth unemployment columns:")
print(youth_unemployment_clean.columns.tolist())

print("NEET columns:")
print(neet.columns.tolist())
```

```python
# ==========================================
# STEP 3 — Prepare Youth Unemployment
# ==========================================

# Convert wide format → long format
# Original youth_unemployment_clean remains unchanged

youth_unemployment_long = youth_unemployment_clean.melt(
    id_vars=[
        "Country Name",
        "Country Code",
        "Indicator Name",
        "Indicator Code"
    ],
    var_name="Year",
    value_name="Youth_Unemployment"
)

# Convert Year to numeric
youth_unemployment_long["Year"] = pd.to_numeric(
    youth_unemployment_long["Year"],
    errors="coerce"
)

# Remove missing unemployment observations
youth_unemployment_long = youth_unemployment_long.dropna(
    subset=["Youth_Unemployment"]
)

# Create a separate analytical dataset
youth_unemployment_final = youth_unemployment_long[
    [
        "Country Name",
        "Country Code",
        "Year",
        "Youth_Unemployment"
    ]
]

# Rename columns to match our other cleaned datasets
youth_unemployment_final = youth_unemployment_final.rename(columns={
    "Country Name": "Country",
    "Country Code": "Country_Code"
})

print("Youth unemployment final shape:", youth_unemployment_final.shape)
print(youth_unemployment_final.head())
```

```python
print(
    "Youth unemployment duplicate country-year rows:",
    youth_unemployment_final.duplicated(
        ["Country_Code", "Year"]
    ).sum()
)
```

```python
# ==========================================
# STEP 5 — Final Compatibility Check
# ==========================================

print("GDP:")
print("  Years:", gdp_clean["Year"].min(), "to", gdp_clean["Year"].max())
print("  Countries:", gdp_clean["Country_Code"].nunique())

print("\nEducation:")
print("  Years:", education_clean["Year"].min(), "to", education_clean["Year"].max())
print("  Countries:", education_clean["Country_Code"].nunique())

print("\nYouth Unemployment:")
print("  Years:", youth_unemployment_final["Year"].min(),"to", youth_unemployment_final["Year"].max())
print("  Countries:", youth_unemployment_final["Country_Code"].nunique())

print("\nNEET:")
print("  Years:", neet_clean["Year"].min(), "to", neet_clean["Year"].max())
print("  Countries:", neet_clean["Country"].nunique())
```

```python
# ==========================================
# STEP 6 — Complete Compatibility Check
# ==========================================

# ------------------------------------------
# COUNTRY OVERLAP
# ------------------------------------------

gdp_countries = set(gdp_clean["Country"].dropna().unique())
education_countries = set(education_clean["Country"].dropna().unique())
youth_countries = set(
    youth_unemployment_final["Country"].dropna().unique()
)
neet_countries = set(neet_clean["Country"].dropna().unique())


common_countries = (
    gdp_countries
    & education_countries
    & youth_countries
    & neet_countries
)

print("========== COUNTRY OVERLAP ==========")

print("GDP countries:", len(gdp_countries))
print("Education countries:", len(education_countries))
print("Youth unemployment countries:", len(youth_countries))
print("NEET countries:", len(neet_countries))

print(
    "\nCountries present in ALL FOUR:",
    len(common_countries)
)
```

```python
# ------------------------------------------
# YEAR OVERLAP
# ------------------------------------------

gdp_years = set(gdp_clean["Year"].dropna().unique())
education_years = set(education_clean["Year"].dropna().unique())
youth_years = set(
    youth_unemployment_final["Year"].dropna().unique()
)
neet_years = set(neet_clean["Year"].dropna().unique())


common_years = (
    gdp_years
    & education_years
    & youth_years
    & neet_years
)

print("\n========== YEAR OVERLAP ==========")

print(
    "GDP:",
    min(gdp_years),
    "to",
    max(gdp_years)
)

print(
    "Education:",
    min(education_years),
    "to",
    max(education_years)
)

print(
    "Youth unemployment:",
    min(youth_years),
    "to",
    max(youth_years)
)

print(
    "NEET:",
    min(neet_years),
    "to",
    max(neet_years)
)

print(
    "\nYears present in ALL FOUR:",
    len(common_years)
)

print(
    "Common period:",
    min(common_years),
    "to",
    max(common_years)
)
```

```python
# ==========================================
# STEP 7 — Find Actual Common Country-Year
# ==========================================

gdp_keys = gdp_clean[["Country", "Year"]].drop_duplicates()

education_keys = education_clean[["Country", "Year"]].drop_duplicates()

youth_keys = youth_unemployment_final[
    ["Country", "Year"]
].drop_duplicates()

neet_keys = neet_clean[
    ["Country", "Year"]
].drop_duplicates()


common_country_year = (
    gdp_keys
    .merge(
        education_keys,
        on=["Country", "Year"],
        how="inner"
    )
    .merge(
        youth_keys,
        on=["Country", "Year"],
        how="inner"
    )
    .merge(
        neet_keys,
        on=["Country", "Year"],
        how="inner"
    )
)

print(
    "Actual common country-year observations:",
    len(common_country_year)
)

print(
    "Countries with all four:",
    common_country_year["Country"].nunique()
)

print(
    "Years with all four:",
    common_country_year["Year"].nunique()
)

print(
    "\nYear range:",
    common_country_year["Year"].min(),
    "to",
    common_country_year["Year"].max()
)

print("\nFirst few rows:")
print(common_country_year.head())
```

```python
from scipy.stats import pearsonr, spearmanr

variables = [
    "GDP_per_capita_PPP",
    "Lower_secondary_completion",
    "Youth_Unemployment",
    "NEET_Rate"
]

for i in range(len(variables)):
    for j in range(i + 1, len(variables)):

        x = project5_data[variables[i]]
        y = project5_data[variables[j]]

        pearson_r, pearson_p = pearsonr(x, y)
        spearman_r, spearman_p = spearmanr(x, y)

        print(f"\n{variables[i]} vs {variables[j]}")
        print(f"Pearson correlation : {pearson_r:.3f}")
        print(f"Spearman correlation: {spearman_r:.3f}")
```
